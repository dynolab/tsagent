Все датасеты реальные, с point-level labels и разнообразными типами аномалий (point spikes по амплитуде, collective/group, trend, frequency-like/ритм).  

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path



def plot_time_series(file_path, start=None, end=None):
    df = pd.read_csv(file_path)
    
    df_slice = df.iloc[start:end]
    
    x = df_slice.index
    y = df_slice['Data']
    labels = df_slice['Label']

    plt.figure(figsize=(12, 6))
    plt.plot(x, y, color='lightblue', linewidth=1.2, zorder=1)
    # plt.scatter(x[labels == 0], y[labels == 0], color='blue', label='Label 0', s=1, zorder=2)
    plt.scatter(x[labels == 1], y[labels == 1], color='red', label='Outliers', s=2, zorder=2)

    plt.title(f'График {file_path} (интервал: {start if start else 'начало'}:{end if end else "конец"})')
    plt.xlabel('Порядковый номер записи')
    plt.ylabel('Значение (Data)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    plt.show()

target_dir = Path('../../../../raw_data')
```

### R1 Combined IT (YAHOO, SMD, IOPS, Exathlon, WSD, NEK)


```python
plot_time_series(f'{target_dir}/R1/YAHOO/Yahoo_A1real_34_data.csv')                                    
plot_time_series(f'{target_dir}/R1/SMD/processed_test_machine-1-1_col_11.csv')                            
plot_time_series(f'{target_dir}/R1/IOPS/KPI-2.csv',0,10000)   

for i in range(3,5):
    try:
        plot_time_series(f'{target_dir}/R1/WSD/WSD_{i}.csv')
    except FileNotFoundError: pass

for i in range(1,1):
    try:
        plot_time_series(f'{target_dir}/R1/NEK/NEK_{i}.csv')
    except FileNotFoundError: pass

plot_time_series(f'{target_dir}/R1/Exathlon/9_3_500000_74.test_col_1.csv') 

```
